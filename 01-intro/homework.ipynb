```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv')
```

```python
df.head()
```

```python
print(len(df))
print(df.shape)
print(df.count)
```

```python
print(df['fuel_type'].unique())
```

```python
q5=df.sort_values(by='fuel_efficiency_mpg', ascending=False).head(3)
print(q5)
```

```python
q6_1=df['horsepower'].median()
print(q6_1)
hasNA=df['horsepower'].isna().any()
print(hasNA)
```

```python
q6_2=df['horsepower'].mode()[0]
print(q6_2)
```

```python
df['horsepower']=df['horsepower'].fillna(q6_2)
q6_4=df['horsepower'].median()
print(q6_4)
```

```python
q7_1_2=df[df['origin']=='Asia'].head(7)
print(q7_1_2)
q7_1_3=q7_1_2[['vehicle_weight', 'model_year']]
print(q7_1_3)
```

```python
import numpy as np
X=q7_1_3.to_numpy()
XTX=X.T @ X
print('top 7 results: ',X)
print('matrix-matrix multiplication: ', XTX)

q7_1_6=np.linalg.inv(XTX)
y=np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w=q7_1_6 @ X.T @ y
print('w is: ',w)
```

```python
print('sum is: ',w[0]+w[1])
```
